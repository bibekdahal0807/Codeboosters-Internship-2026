```python
print("Missing values in each row and column :")
print(df.isnull())

print("Total Missing values in each column :")
print(df.isnull().sum())

print("Total missing values in whole dataset :")
print(df.isnull().sum().sum())
```

```python
print("Students Names :")
print(df['name'])

print(df[['name', 'age', 'department']])
```

```python
print("Departments in dataset:")
print(df['department'].unique())

print("Number of students in each department:")
print(df['department'].value_counts())
```

```python
cs_students = df[df['department'] == 'Computer Science']

print("Number of Computer Science students: ", len(cs_students))

print(cs_students[['name', 'department', 'age', 'gender']])
```

```python
high_attendance = df[df['attendance_percentage'] > 90]

print("Total number of students with more than 90% attendance :", len(high_attendance))

high_attendance[['name', 'department']]
```

```python
df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']

top_student = df.loc[df['total_score'].idxmax()]

print(top_student)
```

```python
dept_avg = df.groupby('department')['total_score'].mean().round(2)
print("Average scores per department:")
print(dept_avg)
```

```python
df_sorted = df.sort_values('total_score', ascending=False)
print(df_sorted[['name','department','total_score']].head(10))
```
