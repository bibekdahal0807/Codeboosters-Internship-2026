```python
# Read the messy_sales_data.csv file into a pandas DataFrame named raw_df
raw_df = pd.read_csv('messy_sales_data.csv')

# Display the first few rows of the raw DataFrame to get an initial look at the data
raw_df.head()
```

```python
# Print the number of rows and columns in the raw DataFrame
print(f"Dataset consists of {raw_df.shape[0]} rows and {raw_df.shape[1]} columns")
# Print the list of all column names in the DataFrame
print(f"Columns : {raw_df.columns.to_list()}")
print("First 5 rows:")

# Display the first 5 rows of the raw DataFrame
raw_df.head()
```

```python
print("=" * 55)
print(" DATA QUALITY DIAGONOSIS REPORT") # Header for the report
print("=" * 55)

# 1. Check for missing values in each column
print("\nMissing values per column:")
print(raw_df.isnull().sum())

# 2. Check for duplicate rows in the DataFrame
print("\nDuplicate rows :", raw_df.duplicated().sum())

# 3. Display data types of each column
print("\nDataTypes :")
print(raw_df.dtypes)

# 4. Display unique values in specific categorical/text columns for initial inspection
print("\nUnique Categories:", raw_df['category'].dropna().unique()) # Unique categories, dropping NaNs
print("\nSample customer name: ", raw_df['customer_name'].dropna().unique()[:8]) # First 8 unique customer names
print("\nsample order values:", raw_df['order_date'].unique()[:6]) # First 6 unique order date values
```

```python
# Create a working copy of the raw DataFrame to perform cleaning operations without altering the original data
df = raw_df.copy()

# Print the shape (number of rows and columns) of the copied DataFrame
print("Copy data of df :", df.shape)
```

```python
print("Total missing values", df.isnull().sum().sum()) # Print the total count of all missing values in the DataFrame

# Fill missing values in the 'order_date' column with a default date '01/01/2019'
df['order_date'].fillna('01/01/2019', inplace = True)

# Fill missing values in the 'category' column with 'unknown'
df['category'].fillna('unknown', inplace = True)

# Fill missing values in the 'customer_name' column with 'Unknown Customer'
df['customer_name'].fillna('Unknown Customer', inplace = True)

# Calculate the median quantity from the 'quantity' column
median_qty = df['quantity'].median()

# Fill missing values in the 'quantity' column with the calculated median quantity
df['quantity'].fillna(median_qty, inplace = True)
```

```python
print(f"Before deuplication: {len(df)} rows") # Print the number of rows before removing duplicates
print(f"Duplicate rows: {df.duplicated().sum()}") # Print the count of duplicate rows

# Show which rows are duplicate by displaying 'order_id', 'customer_name', 'product', 'order_date' for all duplicate entries
print('\nDuplicated rows:')
print(df[df.duplicated(keep=False)][['order_id','customer_name','product','order_date']])

# Remove exact duplicate rows from the DataFrame, keeping only the first occurrence
df.drop_duplicates(inplace=True)

print(f'\nAfter deuplication: {len(df)} rows') # Print the number of rows after removing duplicates
print(f'Rows removed: {len(raw_df) - len(df)}') # Print the count of rows that were removed
```

```python
print('Sample data before passing:')
print(df['order_date'].head(8).tolist()) # Display first 8 order dates before conversion
# Note: some dates are YYYY-MM-DD, some are DD-MM-YYYY, requiring flexible parsing

df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True, # Attempt to parse dates with day first (DD-MM-YYYY) if format is ambiguous
    errors='coerce' # Convert unparseable dates to NaT (Not a Time)
)

print('\nSample data after passing:')
print(df['order_date'].head(8).tolist()) # Display first 8 order dates after conversion

nat_count = df['order_date'].isnull().sum() # Count the number of NaT values (unparseable dates)
print(f'\nUnparseable dates (NaT): {nat_count}')

# Extract date components into new columns
df['year'] = df['order_date'].dt.year # Extract the year
df['month'] = df['order_date'].dt.month # Extract the numerical month
df['month_name'] = df['order_date'].dt.strftime('%B') # Extract the full month name
df['day'] = df['order_date'].dt.day # Extract the day of the month

print('\nSample dates after parsing:')
# Display a sample of the original order_date and the newly extracted components
print(df[['order_date','year','month','month_name','day']].head(8))
```

```python
print('Before standaridazation:', df['customer_name'].unique()[:8]) # Display first 8 unique customer names before standardization

df['customer_name'] = (
    df['customer_name']
    .str.strip() # Remove leading/trailing whitespace (e.g., ' Ram ' to 'Ram')
    .str.title() # Convert names to title case (e.g., 'bibek dahal' to 'Bibek Dahal')
    )

print('\nAfter standardization:', df['customer_name'].unique()[:8]) # Display first 8 unique customer names after standardization

print(f'\n Before Keyboard rows with Electronics category:')
# Create a boolean mask to identify rows where 'product' is 'Keyboard' AND 'category' is 'Electronics'
wrong_mask = (df['product'] == 'Keyboard') & (df['category'] == 'Electronics')
print(df[wrong_mask][['product','category']]) # Display these rows to verify

# Correct the 'category' for these specific rows from 'Electronics' to 'Accessories'
df.loc[wrong_mask, 'category'] = 'Accessories'

print(f'\nAfter fix: Unique categories:', df['category'].unique()) # Display all unique categories after the fix
```

```python
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce') # Convert 'quantity' to numeric, coercing errors to NaN
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce') # Convert 'unit_price' to numeric, coercing errors to NaN

df['revenue'] = df['quantity'] * df['unit_price'] # Calculate 'revenue' by multiplying 'quantity' and 'unit_price'

print('Revenue column created:') # Indicate that the revenue column has been created
# Display a sample of customer name, product, quantity, unit price, and the newly calculated revenue
print(df[['customer_name','product','quantity','unit_price','revenue']].head(5))
# Calculate sum, handling potential NaN values by filling with 0 before summing
total_revenue = df['revenue'].fillna(0).sum() # Fill any NaN revenues with 0 before summing to get total revenue
print(f"\nTotal Revenue across all orders: ${total_revenue:,.0f}") # Print the total revenue, formatted as currency
```

```python
print('='*55)
print(' POST-CLEANING VALIDATION REPORT') # Header for the validation report
print('='*55)
print(f'Orginal rows : {len(raw_df)}') # Print the original number of rows from raw_df
print(f'Cleaned rows : {len(df)}') # Print the number of rows in the cleaned df
print(f'Rows removed : {len(raw_df) - len(df)} (duplicates)') # Calculate and print the number of rows removed (due to duplicates)
print(f'Missing values : {df.isnull().sum().sum()}') # Print the total count of remaining missing values in the cleaned df
print(f'Duplicates : {df.duplicated().sum()}') # Print the count of remaining duplicate rows in the cleaned df
print(f'Date nulls   : {df["order_date"].isnull().sum()}') # Print the count of nulls in the 'order_date' column
print(f'Revenue NaN  : {df["revenue"].isnull().sum()}') # Print the count of NaN values in the 'revenue' column
print(f'Categories   : {sorted(df["category"].unique())}') # Print a sorted list of unique categories
print('='*55)

# Check if the data is considered 'all clean' (no missing values and no duplicates)
all_clean = (
    df.isnull().sum().sum() == 0 and # Condition: no missing values
    df.duplicated().sum() == 0 # Condition: no duplicate rows
)
print(f'DATA IS CLEAN: {all_clean}') # Print the result of the 'all_clean' check
```

```python
product_rev = (
    df.groupby('product')['revenue']
    .sum()
    .reset_index()
    .sort_values('revenue', ascending=False)
)

print('Total revenue by product:')
print(product_rev.to_string(index=False))

category_summary = df.groupby('category').agg(
    total_revenue = ('revenue', 'sum'),
    avg_order_value = ('revenue', 'mean'),
    num_orders = ('order_id', 'count'),
    unique_products = ('product', 'nunique')
).round(2).reset_index()

print('\nCategory Summary:')
print(category_summary.to_string(index=False))
```

```python
df.to_csv('clean_sales_data.csv', index=False)

print('Cleaned data saved to: clean_sales_data.csv')
print(f'Final dataset: {df.shape[0]} rows x {df.shape[1]} columns')
print('\nETL Pipeline for Sales Data: COMPLETE')
print('EXTRACT -> messy_sales_data.csv loaded')
print('TRANSFORM -> cleaned_sales_data.csv saved')
print('LOAD -> clean_sales_data.csv saved')
```
